```python

import mysql.connector
import pandas as pd
import numpy as np

from mysql.connector import Error
from sklearn.cluster import MiniBatchKMeans

# --------------------------------
# CONFIGURATION
# --------------------------------

TABLE_NAME = "labeled_flatdata"

BATCH_SIZE = 10000
N_CLUSTERS = 15

# --------------------------------
# MYSQL CONNECTION
# --------------------------------

def get_connection():
    return mysql.connector.connect(
        host="localhost",
        user="root",
        password="",
        database="marketdata",
    )


# --------------------------------
# OHLC FEATURES
# --------------------------------

features = []

for timeframe in ["H4", "H1", "M15", "M5"]:
    for i in range(1, 21):
        for price in ["open", "high", "low", "close"]:
            features.append(
                f"{timeframe}_{i:02d}_{price}"
            )

assert len(features) == 320

columns_sql = ", ".join(
    f"`{c}`" for c in features
)

# --------------------------------
# MINI BATCH KMEANS
# --------------------------------

model = MiniBatchKMeans(
    n_clusters=N_CLUSTERS,
    batch_size=BATCH_SIZE,
    random_state=42,
    n_init=3,
    max_iter=100,
    reassignment_ratio=0.01
)

connection = get_connection()

if not connection.is_connected():
    raise Exception("MySQL connection failed")

cursor = connection.cursor()

cursor.execute(
    f"SELECT COUNT(*) FROM `{TABLE_NAME}`"
)

total_rows = cursor.fetchone()[0]

print("Total database rows:", total_rows)

# --------------------------------
# PASS 1: INCREMENTAL TRAINING
# --------------------------------

query = f"""
SELECT `id`, {columns_sql}
FROM `{TABLE_NAME}`
ORDER BY `id`
"""

cursor.execute(query)

valid_rows = 0
invalid_rows = 0
batch_number = 0

while True:

    rows = cursor.fetchmany(BATCH_SIZE)

    if not rows:
        break

    data = pd.DataFrame(
        rows,
        columns=["id"] + features
    )

    ids = data["id"].to_numpy()

    X = data[features].to_numpy(
        dtype=np.float32
    )

    valid = np.isfinite(X).all(axis=1)

    X = X[valid]

    if len(X) == 0:
        invalid_rows += len(rows)
        continue

    model.partial_fit(X)

    valid_rows += len(X)
    invalid_rows += len(rows) - len(X)
    batch_number += 1

    print(
        f"Training batch {batch_number} | "
        f"Valid rows: {valid_rows}"
    )

print("\nTraining complete")
print("Valid rows:", valid_rows)
print("Invalid rows:", invalid_rows)

cursor.close()
connection.close()

# --------------------------------
# PASS 2: ASSIGN CLUSTERS
# --------------------------------

connection = get_connection()
cursor = connection.cursor()

cursor.execute(query)

output_file = open(
    "ohlc_cluster_results.csv",
    "w",
    encoding="utf-8"
)

output_file.write("id,cluster\n")

while True:

    rows = cursor.fetchmany(BATCH_SIZE)

    if not rows:
        break

    data = pd.DataFrame(
        rows,
        columns=["id"] + features
    )

    ids = data["id"].to_numpy()

    X = data[features].to_numpy(
        dtype=np.float32
    )

    valid = np.isfinite(X).all(axis=1)

    if not valid.any():
        continue

    labels = model.predict(X[valid])

    result = pd.DataFrame({
        "id": ids[valid],
        "cluster": labels
    })

    result.to_csv(
        output_file,
        index=False,
        header=False
    )

    print(
        "Assigned clusters:",
        len(result)
    )

output_file.close()
cursor.close()
connection.close()

print("\nAll clustering assignments saved!")

# --------------------------------
# SAVE MODEL
# --------------------------------

import joblib

joblib.dump(
    model,
    "ohlc_minibatch_kmeans.pkl"
)

print("Model saved successfully")
```

```python

import pandas as pd
import numpy as np
import joblib

# Load results and trained model
results = pd.read_csv("ohlc_cluster_results.csv")
model = joblib.load("ohlc_minibatch_kmeans.pkl")

# Cluster distribution
distribution = (
    results["cluster"]
    .value_counts()
    .sort_index()
    .rename("rows")
    .to_frame()
)

distribution["percentage"] = (
    distribution["rows"] / len(results) * 100
)

print("\nCLUSTER DISTRIBUTION")
print(distribution.round(2))

print("\nCLUSTER CENTERS")
print(model.cluster_centers_.shape)

# Export cluster distribution
distribution.to_csv("cluster_distribution.csv")

print("\nTotal clusters:", results["cluster"].nunique())
print("Total assigned rows:", len(results))
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import joblib

from sklearn.decomposition import PCA

# Load trained model and cluster assignments
model = joblib.load("ohlc_minibatch_kmeans.pkl")

results = pd.read_csv("ohlc_cluster_results.csv")

# Use a representative sample of 30,000 rows
sample = results.sample(
    n=min(30000, len(results)),
    random_state=42
)

# Fetch corresponding original OHLC rows
connection = get_connection()
cursor = connection.cursor()

ids = sample["id"].tolist()

# Fetch in manageable chunks
features = []

for timeframe in ["H4", "H1", "M15", "M5"]:
    for i in range(1, 21):
        for price in ["open", "high", "low", "close"]:
            features.append(
                f"{timeframe}_{i:02d}_{price}"
            )

columns_sql = ", ".join(
    f"`{c}`" for c in features
)

frames = []

for start in range(0, len(ids), 1000):

    chunk = ids[start:start + 1000]

    placeholders = ",".join(["%s"] * len(chunk))

    query = f"""
    SELECT `id`, {columns_sql}
    FROM labeled_flatdata
    WHERE id IN ({placeholders})
    """

    cursor.execute(query, chunk)

    rows = cursor.fetchall()

    frames.append(
        pd.DataFrame(
            rows,
            columns=["id"] + features
        )
    )

cursor.close()
connection.close()

data = pd.concat(frames, ignore_index=True)

# Preserve cluster assignment by database ID
data = data.merge(
    sample[["id", "cluster"]],
    on="id"
)

X = data[features].to_numpy(dtype="float32")

# PCA
pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(X)

# Visualization
plt.figure(figsize=(13, 9))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=data["cluster"],
    cmap="tab20",
    s=5,
    alpha=0.5
)

plt.colorbar(scatter, label="Cluster")

plt.title("OHLC Pattern Clustering - PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.tight_layout()
plt.savefig("ohlc_pca_clusters.png", dpi=200)
plt.show()

print(
    "Explained variance:",
    pca.explained_variance_ratio_.sum()
)
```
