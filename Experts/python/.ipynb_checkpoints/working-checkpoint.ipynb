```python
import pandas as pd

def flatten_to_single_row(df_input, prefix=""):
    """
    Flattens a multi-row DataFrame into a single-row DataFrame.
    Appends the row index (0, 1, 2...) to each column name.
    """
    if df_input.empty:
        return pd.DataFrame()
    
    # 1. Unstack the dataframe (converts rows to a Series with a MultiIndex)
    unstacked = df_input.unstack()
    
    # 2. Rename index levels: level 0 is column name, level 1 is row index
    # Format: colname_0, colname_1, etc.
    new_cols = [f"{prefix}{col}_{idx}" for col, idx in unstacked.index]
    
    # 3. Create a single-row DataFrame
    single_row_df = pd.DataFrame([unstacked.values], columns=new_cols)
    return single_row_df

# --- Flatten each of your 3 DataFrames ---
df_raw_1row = flatten_to_single_row(df_raw, prefix="raw_")
df_candles_1row = flatten_to_single_row(df_candles, prefix="group_")
df_market_1row = flatten_to_single_row(df_market, prefix="market_")

# --- Combine all 3 into ONE single master row ---
master_single_row = pd.concat([df_raw_1row, df_candles_1row, df_market_1row], axis=1)

# Check shape (should be 1 row, N columns)
print("Shape of final DataFrame:", master_single_row.shape)
print("\nFirst 10 column names generated:")
print(master_single_row.columns[:10].tolist())

# View full single-row dataset
print("\nFinal Single Row DataFrame:")
print(master_single_row)
```

```python
import pandas as pd

# 1. Access row1 dictionary from 'samples'
row1 = df['samples'].iloc[0]

# 2. Convert nested lists to DataFrames
df_raw = pd.json_normalize(row1['raw_data'])
df_candles = pd.json_normalize(row1['continuous_candle_group'])
df_market = pd.json_normalize(row1['market_analysis'])

# 3. Helper function to unstack rows into single row columns
def flatten_to_single_row(df_input, prefix=""):
    if df_input.empty:
        return pd.DataFrame()
    unstacked = df_input.unstack()
    new_cols = [f"{prefix}{col}_{idx}" for col, idx in unstacked.index]
    return pd.DataFrame([unstacked.values], columns=new_cols)

# 4. Flatten all sub-DataFrames
df_raw_1row = flatten_to_single_row(df_raw, prefix="raw_")
df_candles_1row = flatten_to_single_row(df_candles, prefix="group_")
df_market_1row = flatten_to_single_row(df_market, prefix="market_")

# 5. Concatenate side-by-side into 1 single row
master_single_row = pd.concat([df_raw_1row, df_candles_1row, df_market_1row], axis=1)

# 6. Export to CSV (index=False avoids adding an unneeded row index)
master_single_row.to_csv("flattened_ml_dataset.csv", index=False)

print("Dataset exported successfully!")
```

```python
import pandas as pd

# Extract row1 dictionary from 'samples'
row1 = df['samples'].iloc[0]

# Load nested dicts into DataFrames
df_raw = pd.json_normalize(row1['raw_data'])
df_candles = pd.json_normalize(row1['continuous_candle_group'])
df_market = pd.json_normalize(row1['market_analysis'])

def flatten_simple_numbered(df_input, prefix="raw"):
    """
    Flattens rows into clean, concise numbered column headers.
    Format: time_1, open_1, high_1 ... time_2, open_2 ...
    """
    if df_input.empty:
        return pd.DataFrame()
    
    dict_row = {}
    for idx, row in df_input.iterrows():
        # Simple suffix: _1, _2, _3...
        num = idx + 1
            
        for col in df_input.columns:
            # If using prefixes for distinct sections, e.g., raw_time_1 vs group_time_1
            col_name = f"{prefix}_{col}_{num}" if prefix else f"{col}_{num}"
            dict_row[col_name] = row[col]
            
    return pd.DataFrame([dict_row])

# Process sections (use prefix="" if you want purely time_1, open_1...)
df_raw_1row = flatten_simple_numbered(df_raw, prefix="")
df_candles_1row = flatten_simple_numbered(df_candles, prefix="group")
df_market_1row = flatten_simple_numbered(df_market, prefix="market")

# Combine into a single row
master_single_row = pd.concat([df_raw_1row, df_candles_1row, df_market_1row], axis=1)

# Export to CSV
master_single_row.to_csv("simple_numbered_dataset.csv", index=False)

print("Exported! Sample column headers:")
print(list(master_single_row.columns[:10]))
```

```python
import pandas as pd

def flatten_simple_numbered(df_input, prefix=""):
    """
    Flattens a multi-row DataFrame into a single-row DataFrame.
    Format: time_1, open_1, high_1 ... time_2, open_2 ...
    """
    if df_input.empty:
        return pd.DataFrame()
    
    dict_row = {}
    for idx, row in df_input.iterrows():
        num = idx + 1
        for col in df_input.columns:
            col_name = f"{prefix}_{col}_{num}" if prefix else f"{col}_{num}"
            dict_row[col_name] = row[col]
            
    return pd.DataFrame([dict_row])

def process_sample_row(sample_dict):
    """Processes a single sample dictionary into a 1-row flattened DataFrame."""
    df_raw = pd.json_normalize(sample_dict.get('raw_data', []))
    df_candles = pd.json_normalize(sample_dict.get('continuous_candle_group', []))
    df_market = pd.json_normalize(sample_dict.get('market_analysis', []))
    
    df_raw_1row = flatten_simple_numbered(df_raw, prefix="")
    df_candles_1row = flatten_simple_numbered(df_candles, prefix="group")
    df_market_1row = flatten_simple_numbered(df_market, prefix="market")
    
    return pd.concat([df_raw_1row, df_candles_1row, df_market_1row], axis=1)

# 1. Process all samples in df['samples']
all_flattened_rows = [process_sample_row(sample) for sample in df['samples']]

# 2. Stack all rows together into one master dataset
final_ml_dataset = pd.concat(all_flattened_rows, ignore_index=True)

# 3. Export to CSV
final_ml_dataset.to_csv("full_flattened_ml_dataset.csv", index=False)

print(f"Dataset successfully exported! Shape: {final_ml_dataset.shape}")
print("\nFirst 10 columns:")
print(list(final_ml_dataset.columns[:10]))
```
