```python
import pandas as pd

# 1. Access row1 dictionary from 'samples'
row1 = df['samples'].iloc[0]

# 2. Convert nested lists to DataFrames
df_raw = pd.json_normalize(row1['raw_data'])
df_candles = pd.json_normalize(row1['continuous_candle_group'])
df_market = pd.json_normalize(row1['market_analysis'])

# 3. Helper function to unstack rows into single row columns
def flatten_to_single_row(df_input, prefix=""):
    if df_input.empty:
        return pd.DataFrame()
    unstacked = df_input.unstack()
    new_cols = [f"{prefix}{col}_{idx}" for col, idx in unstacked.index]
    return pd.DataFrame([unstacked.values], columns=new_cols)

# 4. Flatten all sub-DataFrames
df_raw_1row = flatten_to_single_row(df_raw, prefix="raw_")
df_candles_1row = flatten_to_single_row(df_candles, prefix="group_")
df_market_1row = flatten_to_single_row(df_market, prefix="market_")

# 5. Concatenate side-by-side into 1 single row
master_single_row = pd.concat([df_raw_1row, df_candles_1row, df_market_1row], axis=1)

# 6. Export to CSV (index=False avoids adding an unneeded row index)
master_single_row.to_csv("flattened_ml_dataset.csv", index=False)

print("Dataset exported successfully!")
```

```python
# ============================================================
# JUPYTERLAB: JSON -> FLATTEN -> SESSION -> DJANGO DATABASE
# ============================================================

import os
import sys
from pathlib import Path
from datetime import datetime

import pandas as pd
import numpy as np


# ============================================================
# 1. DJANGO SETUP
# ============================================================

PROJECT_DIR = Path(
    r"D:\Software\Projects\webstation\django"
)

if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

os.environ.setdefault(
    "DJANGO_SETTINGS_MODULE",
    "webstation.settings"
)

import django
django.setup()

from django.db import transaction
from forexdata.models import MLDataset


# ============================================================
# 2. JSON FILE
# ============================================================

file_path = Path(
    r"C:\Users\nishw\AppData\Roaming\MetaQuotes\Tester"
    r"\D0E8209F77C8CF37AD8BF550E51FF075"
    r"\Agent-127.0.0.1-3000\MQL5\Files"
    r"\EURUSD_PERIOD_CURRENT_ml_dataset.json"
)


# ============================================================
# 3. LOAD JSON
# ============================================================

df = pd.read_json(file_path)

print("JSON loaded")
print("Original shape:", df.shape)

if "samples" not in df.columns:
    raise ValueError(
        "ERROR: 'samples' column not found in JSON dataset."
    )


# ============================================================
# 4. FLATTEN FUNCTION
# ============================================================

def flatten_simple_numbered(df_input, prefix=""):

    if df_input.empty:
        return pd.DataFrame()

    flattened = {}

    for idx, row in df_input.iterrows():

        num = idx + 1

        for col in df_input.columns:

            if prefix:
                col_name = f"{prefix}_{col}_{num}"
            else:
                col_name = f"{col}_{num}"

            flattened[col_name] = row[col]

    return pd.DataFrame([flattened])


# ============================================================
# 5. PROCESS ONE SAMPLE
# ============================================================

def process_sample_row(sample_dict):

    df_raw = pd.json_normalize(
        sample_dict.get("raw_data", [])
    )

    df_candles = pd.json_normalize(
        sample_dict.get("continuous_candle_group", [])
    )

    df_market = pd.json_normalize(
        sample_dict.get("market_analysis", [])
    )

    df_raw_1row = flatten_simple_numbered(
        df_raw,
        prefix=""
    )

    df_candles_1row = flatten_simple_numbered(
        df_candles,
        prefix="group"
    )

    df_market_1row = flatten_simple_numbered(
        df_market,
        prefix="market"
    )

    return pd.concat(
        [
            df_raw_1row,
            df_candles_1row,
            df_market_1row
        ],
        axis=1
    )


# ============================================================
# 6. PROCESS ALL SAMPLES
# ============================================================

all_flattened_rows = []

for sample_number, sample in enumerate(
    df["samples"],
    start=1
):

    try:

        flattened = process_sample_row(sample)

        if not flattened.empty:
            all_flattened_rows.append(flattened)

    except Exception as e:

        print(
            f"ERROR processing sample {sample_number}: {e}"
        )


if not all_flattened_rows:
    raise ValueError(
        "No flattened samples were generated."
    )


final_ml_dataset = pd.concat(
    all_flattened_rows,
    ignore_index=True
)


# ============================================================
# 7. ADD SESSION COLUMN
# ============================================================

current_hour = datetime.now().hour

session_value = f"session_{current_hour}"

final_ml_dataset.insert(
    0,
    "session",
    session_value
)


# ============================================================
# 8. CLEAN VALUES FOR JSON
# ============================================================

def make_json_safe(value):

    if value is None:
        return None

    try:
        if pd.isna(value):
            return None
    except Exception:
        pass

    if isinstance(value, pd.Timestamp):
        return value.isoformat()

    if isinstance(value, datetime):
        return value.isoformat()

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    return value


# ============================================================
# 9. EXPORT CSV
# ============================================================

csv_path = PROJECT_DIR / "full_flattened_ml_dataset.csv"

final_ml_dataset.to_csv(
    csv_path,
    index=False
)

print()
print("=" * 60)
print("CSV CREATED")
print("=" * 60)
print("Path:", csv_path)
print("Shape:", final_ml_dataset.shape)
print("Session:", session_value)


# ============================================================
# 10. PREPARE DJANGO RECORDS
# ============================================================

records = []

for _, row in final_ml_dataset.iterrows():

    data = {}

    for column in final_ml_dataset.columns:

        if column == "session":
            continue

        data[column] = make_json_safe(
            row[column]
        )

    records.append(
        MLDataset(
            session=session_value,
            data=data
        )
    )


# ============================================================
# 11. INSERT INTO DJANGO DATABASE
# ============================================================

with transaction.atomic():

    MLDataset.objects.bulk_create(
        records,
        batch_size=500
    )


# ============================================================
# 12. VERIFY
# ============================================================

total_records = MLDataset.objects.count()

session_records = MLDataset.objects.filter(
    session=session_value
).count()


print()
print("=" * 60)
print("DJANGO DATABASE INSERT COMPLETE")
print("=" * 60)
print("Inserted:", len(records))
print("Session:", session_value)
print("Records in this session:", session_records)
print("Total database records:", total_records)
print("=" * 60)
```
