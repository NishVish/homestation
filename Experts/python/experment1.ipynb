```python
import pandas as pd
import mysql.connector

# =========================
# 1. Read CSV
# =========================
df = pd.read_csv(r"dataset\1min.csv")

print("Rows:", len(df))
print("Columns:", df.columns.tolist())


# =========================
# 2. Connect to MySQL
# =========================
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="marketdata",
    port=3306
)

cursor = conn.cursor()


# =========================
# 3. Create table automatically
# =========================

columns = df.columns.tolist()

# Make every CSV column VARCHAR first.
# This avoids datatype problems during import.
column_definitions = ", ".join(
    f"`{col}` TEXT" for col in columns
)

create_sql = f"""
CREATE TABLE IF NOT EXISTS `1min` (
    {column_definitions}
)
"""

cursor.execute(create_sql)

print("Table `1min` is ready.")


# =========================
# 4. Insert ALL CSV columns
# =========================

column_names = ", ".join(
    f"`{col}`" for col in columns
)

placeholders = ", ".join(
    ["%s"] * len(columns)
)

insert_sql = f"""
INSERT INTO `1min` ({column_names})
VALUES ({placeholders})
"""

# Convert NaN to None for MySQL
df = df.where(pd.notnull(df), None)

data = list(
    df.itertuples(index=False, name=None)
)

cursor.executemany(insert_sql, data)

conn.commit()

print(f"Inserted {cursor.rowcount} rows.")


# =========================
# 5. Close connection
# =========================

cursor.close()
conn.close()

print("Done.")
```
