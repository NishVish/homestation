```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# ============================================================
# STEP 1 — LOAD JSON
# ============================================================

file_path = r"C:\Users\nishw\AppData\Roaming\MetaQuotes\Tester\D0E8209F77C8CF37AD8BF550E51FF075\Agent-127.0.0.1-3000\MQL5\Files\EURUSD_PERIOD_CURRENT_ml_dataset.json"

df = pd.read_json(file_path)

print(df.head())
print(df.info())


# ============================================================
# STEP 2 — FLATTEN DATA
# ============================================================

row1 = df["samples"].iloc[0]

df_raw = pd.json_normalize(row1["raw_data"])
df_candles = pd.json_normalize(row1["continuous_candle_group"])
df_market = pd.json_normalize(row1["market_analysis"])


def flatten_to_single_row(df_input, prefix=""):
    if df_input.empty:
        return pd.DataFrame()

    unstacked = df_input.unstack()

    new_cols = [
        f"{prefix}{col}_{idx}"
        for col, idx in unstacked.index
    ]

    return pd.DataFrame(
        [unstacked.values],
        columns=new_cols
    )


df_raw_1row = flatten_to_single_row(df_raw, prefix="raw_")
df_candles_1row = flatten_to_single_row(df_candles, prefix="group_")
df_market_1row = flatten_to_single_row(df_market, prefix="market_")

master_single_row = pd.concat(
    [
        df_raw_1row,
        df_candles_1row,
        df_market_1row
    ],
    axis=1
)

master_single_row.to_csv(
    "flattened_ml_dataset.csv",
    index=False
)

print("Dataset exported successfully!")


# ============================================================
# STEP 3 — MYSQL DATABASE
# ============================================================

DB_USER = "root"
DB_PASSWORD = ""
DB_HOST = "localhost"
DB_PORT = 3306
DB_NAME = "trading_ml"
TABLE_NAME = "eurusd_ml_dataset"

encoded_password = quote_plus(DB_PASSWORD)

engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{encoded_password}@"
    f"{DB_HOST}:{DB_PORT}/{DB_NAME}"
)


# ============================================================
# CREATE DATABASE
# ============================================================

with engine.connect() as conn:
    print("Connected to MySQL successfully!")


# ============================================================
# INSERT DATA
# ============================================================

master_single_row.to_sql(
    name=TABLE_NAME,
    con=engine,
    if_exists="append",
    index=False
)

print(f"Inserted {len(master_single_row)} row(s) into {TABLE_NAME}")

# Verify
check = pd.read_sql(
    f"SELECT * FROM `{TABLE_NAME}` LIMIT 5",
    engine
)

print(check)
```
