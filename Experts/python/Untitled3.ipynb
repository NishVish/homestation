```python
import duckdb

DB_PATH = r"C:\Users\nishw\AppData\Roaming\MetaQuotes\Terminal\D0E8209F77C8CF37AD8BF550E51FF075\MQL5\Experts\python\annotation_data.db"

# Connect in read-write mode to modify the schema
con = duckdb.connect(DB_PATH)

# Recreate the table with an auto-incrementing 'id' column as PRIMARY KEY
con.execute("""
    CREATE OR REPLACE TABLE market_data AS 
    SELECT 
        ROW_NUMBER() OVER () AS id,
        * 
    FROM market_data;
""")

print("Successfully added permanent 'id' column to 'market_data'!")
con.close()
```
